Task 2: Exploratory Data Analysis

2.1 Transaction trends over the years for each state

```python
TxnTrends_each_state = StateTxn_Users.groupby(['State', 'Year'])['Transactions'].sum().reset_index()
TxnTrends_each_state
```

```python
Top5_highestTxn = TxnTrends_each_state.groupby('State')['Transactions'].sum().sort_values(ascending = False).head(5).reset_index()
Top5_highestTxn.index = np.arange(1, len(Top5_highestTxn) + 1)
print('Top 5 States with highest transaction volumes are : ')
Top5_highestTxn
```

```python
Top5_lowestTxn = TxnTrends_each_state.groupby('State')['Transactions'].sum().sort_values(ascending = True).head(5).reset_index()
Top5_lowestTxn.index = np.arange(1, len(Top5_lowestTxn) + 1)
print('Top 5 States with lowest transaction volumes are : ')
Top5_lowestTxn
```

Task 4 : Data merging and advanced analysis

4.1 Ratio of registered users to population by state

```python
merged_SU_DD = pd.merge(StateTxn_Users, District_Demographics, on='State', how='inner')
merged_SU_DD
```

```python
State_RU_P = merged_SU_DD.groupby('State').agg({'Registered Users': 'max','Population': 'sum'}).reset_index()
State_RU_P['Ratio_U/P'] = State_RU_P['Registered Users'] / State_RU_P['Population']
State_RU_P
```

```python
plt.figure(figsize = (16,9))
plt.bar(State_RU_P['State'], State_RU_P['Ratio_U/P'], color = 'orange')
plt.xlabel('State')
plt.ylabel('Ratio_U/P')
plt.xticks(rotation = 90)
plt.title('Ratio of registered users to population for each state')
```

4.2 Correlation of population density with transaction volume

```python
merged_DU_DD = pd.merge(DistrictTxn_Users, District_Demographics, on = 'District', how = 'inner')
merged_DU_DD
```

```python
District_PD_TV = merged_DU_DD.groupby('District').agg({'Density' : 'max', 'Transactions' : 'sum' }).reset_index()
District_PD_TV
```

```python
Corr_PD_TV = District_PD_TV['Density'].corr(District_PD_TV['Transactions'])
Corr_PD_TV
```

4.3 Average transaction amount per user

```python
State_TA_RU = DistrictTxn_Users.groupby('State').agg({'Amount (INR)':'sum', 'Registered Users':'sum'}).reset_index()
State_TA_RU['AvgTxnAmt_Per_User'] = State_TA_RU['Amount (INR)']/State_TA_RU['Registered Users']
State_TA_RU
```

```python
Top_5_Lowest_AvgTxnAmt_Per_User = State_TA_RU.sort_values(by = 'AvgTxnAmt_Per_User', ascending = True).head(5).reset_index(drop = True)
Top_5_Lowest_AvgTxnAmt_Per_User.set_index(np.arange(1, len(Top_5_Lowest_AvgTxnAmt_Per_User) + 1), inplace = True)
print("Top 5 States with Lowest Average Transaction Amount Per User are : ")
Top_5_Lowest_AvgTxnAmt_Per_User
```

Task 6: Insights and conclusions

6.1 Trends/Patterns in the transaction data

```python
Top5_Highest_TxnStates = Top5_highestTxn['State'].to_list()
Top5_Lowest_TxnStates = Top5_lowestTxn['State'].to_list()
```

```python
plt.figure(figsize = (16,9))
for state in Top5_Highest_TxnStates :
    subset = StateTxn_Users[StateTxn_Users['State'] == state]
    plt.plot(subset['Year'], subset['Transactions'], marker = 'o', label = state)
plt.xlabel('Year')
plt.ylabel('Transactions')
plt.title('Transaction trends over years for Top 5 highest transaction states')
plt.legend()
```

📊 Trend Overview
Consistent Growth in Top 5 Highest Transaction States: All five states—Maharashtra, Karnataka, Telangana, Andhra Pradesh, and Rajasthan—show a steady year-on-year increase in transaction volumes from 2018 to 2021.

Accelerated Uptick: The growth rate appears to accelerate post-2019, likely reflecting increased digital adoption, policy pushes (like UPI expansion), and pandemic-driven digitization.

🥇 State-wise Insights
1. Maharashtra - Sharp rise - In 2021 its position is Highest.
2. Karnataka - Strong growth - In 2021 its position is 2nd Highest.
3. Telangana - Moderate rise - In 2021 its position is Mid-tier.
4. Andhra Pradesh - Gradual increase - In 2021 its position is slightly below Telangana.
5. Rajasthan - Slowest growth in the top 5 highest transaction states - In 2021 its position is lowest. 

```python
plt.figure(figsize = (16,9))
for state in Top5_Lowest_TxnStates :
    subset = StateTxn_Users[StateTxn_Users['State'] == state]
    plt.plot(subset['Year'], subset['Transactions'], marker = 'o', label = state)
plt.xlabel('Year')
plt.ylabel('Transactions')
plt.title('Transaction trends over years for Top 5 lowest transaction states')
plt.legend()
```

📉 Trend Description
🔍 General Pattern
All five regions—Lakshadweep, Andaman & Nicobar Islands, Ladakh, Mizoram, and Meghalaya—show a positive upward trend in transaction volumes over the years.

The overall scale of transactions is significantly lower compared to high-performing states, but the growth trajectory is encouraging.

📊 Key Insights


Consistent Growth in High-Transaction States The plotted lines suggest that states with the highest transaction volumes show steady year-on-year growth. This indicates strong adoption of digital payments and increasing user trust in platforms like PhonePe.

Regional Concentration of Digital Activity The fact that only a handful of states dominate the transaction charts highlights regional disparities. Economic activity and digital penetration are concentrated in these states, which could be linked to higher urbanization, better infrastructure, and stronger financial ecosystems.

User Base Expansion Driving Transactions The upward trajectory of transactions aligns with the growing number of users in these states. This suggests that PhonePe’s expansion strategy is effectively capturing new users and encouraging repeat usage.

Potential Market Saturation Signals If the growth curves begin to flatten in future years, it may indicate saturation in these top states. PhonePe might then need to focus on lower-transaction states to sustain overall growth.

Strategic Implications

For business strategy: Prioritize customer engagement and retention in high-performing states while investing in awareness campaigns in low-performing ones.

For policy makers: Address digital divide issues by improving infrastructure and financial literacy in lagging regions.

For investors: The growth trajectory in top states signals strong revenue potential, but diversification into emerging states is key for long-term sustainability.
